## Valorant API

```python
url = f'https://valorant-api.com/v1/agents'
response = requests.get(url)
lista_agente = response.json()
lista_agente
base_agentes = pd.DataFrame(lista_agente['data'])
base_agentes
```

```python
AGENTE = base_agentes.loc[base_agentes['displayName'] == 'Sova', 'uuid'].values[0]

url = f'https://valorant-api.com/v1/agents/{AGENTE}'
response = requests.get(url)
status_sova = response.json()
status_sova = pd.DataFrame([status_sova['data']])
status_sova
```
